```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import squarify
```

```python
df = pd.read_csv("age_salary_hours.csv")
```

```python
print(df.columns.tolist())
```

```python
edu_counts = df['Education'].value_counts(normalize=True) * 100
labels = [f"{edu}\n({pct:.1f}%)" for edu, pct in edu_counts.items()]

plt.figure(dpi=200)
squarify.plot(sizes=edu_counts.values,
              label=labels,
              color=sns.color_palette('colorblind', len(edu_counts)))
plt.axis('off')
plt.title('Degrees')
plt.show()
```

```python
df_filtered = df[df['Age'] < 65]
```

```python
fig, axes = plt.subplots(2, 1, figsize=(10, 10))


sns.violinplot(data=df_filtered, x='Education', y='Annual Salary',
               ax=axes[0], hue='Education', legend=False, palette='colorblind')
axes[0].set_title('Impact of Education on Annual Salary and Weekly Working Hours')
axes[0].set_xlabel('Education')
axes[0].set_ylabel('Annual Salary')
axes[0].tick_params(axis='x', rotation=15)
sns.violinplot(data=df_filtered, x='Education', y='Weekly hours',
               ax=axes[1], hue='Education', legend=False, palette='colorblind')
axes[1].set_xlabel('Education')
axes[1].set_ylabel('Weekly Hours')
axes[1].tick_params(axis='x', rotation=15)

plt.tight_layout()
plt.show()
```
